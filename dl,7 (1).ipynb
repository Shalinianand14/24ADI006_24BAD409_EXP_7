```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os

# =========================================================
# EXPT DETAILS
# =========================================================
print("EXPT NO: 7 - GAN")
print("Student Roll No: 24BAD409")
print("Student Name: Shalini A")
print("TensorFlow Version:", tf.__version__)

# =========================================================
# PART A: DATASET PREPARATION AND PREPROCESSING
# =========================================================
(X_train, _), (_, _) = tf.keras.datasets.mnist.load_data()
X_train = X_train[:20000]
print("\nNumber of training images:", X_train.shape[0])
print("Image dimensions:", X_train.shape[1:])

# Display real images
plt.figure(figsize=(10, 5))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.axis("off")
    plt.title("Real Image")
plt.tight_layout()
plt.show()
# Normalize pixels to [-1, 1]
X_train = X_train.astype("float32")
X_train = (X_train - 127.5) / 127.5
# Add channel dimension
X_train = np.expand_dims(X_train, axis=-1)
print("Minimum pixel value:", X_train.min())
print("Maximum pixel value:", X_train.max())
print("Final dataset shape:", X_train.shape)
# Create TensorFlow dataset
BATCH_SIZE = 256
train_dataset = (
    tf.data.Dataset
    .from_tensor_slices(X_train)
    .shuffle(20000)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE))
print("Dataset pipeline created successfully.")

# =========================================================
# PART B: GENERATOR
# =========================================================
NOISE_DIM = 100
generator = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(NOISE_DIM,)),
    tf.keras.layers.Dense(
        7 * 7 * 256,
        use_bias=False
    ),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.LeakyReLU(),
    tf.keras.layers.Reshape(
        (7, 7, 256)
    ),
    tf.keras.layers.Conv2DTranspose(
        128,
        kernel_size=5,
        strides=1,
        padding="same",
        use_bias=False
    ),

    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.LeakyReLU(),
    tf.keras.layers.Conv2DTranspose(
        64,
        kernel_size=5,
        strides=2,
        padding="same",
        use_bias=False
    ),

    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.LeakyReLU(),
    tf.keras.layers.Conv2DTranspose(
        1,
        kernel_size=5,
        strides=2,
        padding="same",
        use_bias=False,
        activation="tanh"
    )
], name="Generator")
print("\nGENERATOR MODEL SUMMARY")
generator.summary()

# Generate image before training
noise = tf.random.normal([1, NOISE_DIM])
generated_image = generator(
    noise,
    training=False
)
print("Generated image shape:", generated_image.shape)
plt.figure(figsize=(4, 4))
plt.imshow(
    generated_image[0, :, :, 0],
    cmap="gray"
)
plt.axis("off")
plt.title("Generated Image Before Training")
plt.show()

# =========================================================
# PART C: DISCRIMINATOR
# =========================================================
discriminator = tf.keras.Sequential([
    tf.keras.layers.Input(
        shape=(28, 28, 1)
    ),
    tf.keras.layers.Conv2D(
        64,
        kernel_size=5,
        strides=2,
        padding="same"
    ),
    tf.keras.layers.LeakyReLU(),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Conv2D(
        128,
        kernel_size=5,
        strides=2,
        padding="same"),

    tf.keras.layers.LeakyReLU(),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(1)
], name="Discriminator")

print("\nDISCRIMINATOR MODEL SUMMARY")
discriminator.summary()
# Test discriminator
decision = discriminator(
    generated_image,
    training=False
)
print("Discriminator output:", decision.numpy())
# Binary cross entropy
cross_entropy = tf.keras.losses.BinaryCrossentropy(
    from_logits=True
)
# Discriminator loss
def discriminator_loss(real_output, fake_output):
    real_loss = cross_entropy(
        tf.ones_like(real_output),
        real_output
    )
    fake_loss = cross_entropy(
        tf.zeros_like(fake_output),
        fake_output
    )
    return real_loss + fake_loss

# Generator loss
def generator_loss(fake_output):
    return cross_entropy(
        tf.ones_like(fake_output),
        fake_output)

# Optimizers
generator_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0002,
    beta_1=0.5)

discriminator_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0002,
    beta_1=0.5)

# =========================================================
# PART D: GAN TRAINING
# =========================================================
EPOCHS = 5
NUM_EXAMPLES_TO_GENERATE = 16
seed = tf.random.normal(
    [NUM_EXAMPLES_TO_GENERATE, NOISE_DIM])
generator_losses = []
discriminator_losses = []

@tf.function
def train_step(images):
    noise = tf.random.normal(
        [tf.shape(images)[0], NOISE_DIM])

    with tf.GradientTape() as gen_tape, \
         tf.GradientTape() as disc_tape:
        # Generate fake images
        generated_images = generator(
            noise,
            training=True)

        # Discriminator predictions
        real_output = discriminator(
            images,
            training=True)

        fake_output = discriminator(
            generated_images,
            training=True)

        # Calculate losses
        gen_loss = generator_loss(
            fake_output)

        disc_loss = discriminator_loss(
            real_output,
            fake_output)

    # Generator gradients
    gradients_of_generator = gen_tape.gradient(
        gen_loss,
        generator.trainable_variables)

    # Discriminator gradients
    gradients_of_discriminator = disc_tape.gradient(
        disc_loss,
        discriminator.trainable_variables)

    # Update Generator
    generator_optimizer.apply_gradients(
        zip(
            gradients_of_generator,
            generator.trainable_variables)
    )

    # Update Discriminator
    discriminator_optimizer.apply_gradients(
        zip(
            gradients_of_discriminator,
            discriminator.trainable_variables)
    )

    return gen_loss, disc_loss

# =========================================================
# DISPLAY GENERATED IMAGES
# =========================================================
def generate_and_display_images(
    model,
    epoch,
    test_input):
    predictions = model(
        test_input,
        training=False)
    plt.figure(figsize=(8, 8))
    for i in range(predictions.shape[0]):
        plt.subplot(4, 4, i + 1)
        plt.imshow(
            predictions[i, :, :, 0] * 0.5 + 0.5,
            cmap="gray")

        plt.axis("off")
    plt.suptitle(
        "Generated Images - Epoch " + str(epoch)
    )
    plt.tight_layout()
    plt.show()

# =========================================================
# TRAINING LOOP
# =========================================================
print("\nGAN TRAINING STARTED")
for epoch in range(1, EPOCHS + 1):
    epoch_generator_loss = []
    epoch_discriminator_loss = []
    for image_batch in train_dataset:
        gen_loss, disc_loss = train_step(
            image_batch)

        epoch_generator_loss.append(
            float(gen_loss))

        epoch_discriminator_loss.append(
            float(disc_loss))

    avg_generator_loss = np.mean(
        epoch_generator_loss
    )

    avg_discriminator_loss = np.mean(
        epoch_discriminator_loss
    )

    generator_losses.append(
        avg_generator_loss
    )

    discriminator_losses.append(
        avg_discriminator_loss
    )

    print(
        "Epoch {}/{} - Generator Loss: {:.4f} - Discriminator Loss: {:.4f}".format(
            epoch,
            EPOCHS,
            avg_generator_loss,
            avg_discriminator_loss))

    generate_and_display_images(
        generator,
        epoch,
        seed)

# =========================================================
# LOSS GRAPH
# =========================================================
plt.figure(figsize=(10, 5))
plt.plot(
    generator_losses,
    label="Generator Loss")

plt.plot(
    discriminator_losses,
    label="Discriminator Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GAN Training Loss")
plt.legend()
plt.grid(True)
plt.show()

# =========================================================
# REAL VS SYNTHETIC IMAGES
# =========================================================

real_images = X_train[:16]
noise = tf.random.normal(
    [16, NOISE_DIM])
fake_images = generator(
    noise,
    training=False)

# Real images
plt.figure(figsize=(10, 8))
for i in range(16):
    plt.subplot(4, 4, i + 1)
    plt.imshow(
        real_images[i, :, :, 0] * 0.5 + 0.5,
        cmap="gray"
    )
    plt.axis("off")
plt.suptitle("Real MNIST Images")
plt.tight_layout()
plt.show()

# Synthetic images
plt.figure(figsize=(10, 8))
for i in range(16):
    plt.subplot(4, 4, i + 1)
    plt.imshow(
        fake_images[i, :, :, 0] * 0.5 + 0.5,
        cmap="gray"
    )
    plt.axis("off")
plt.suptitle(
    "Synthetic Images Generated by GAN"
)
plt.tight_layout()
plt.show()

# =========================================================
# SAVE GENERATED IMAGES
# =========================================================
os.makedirs(
    "generated_images",
    exist_ok=True
)
for i in range(16):
    image = (
        fake_images[i, :, :, 0] * 0.5 + 0.5
    )
    plt.imsave(
        "generated_images/generated_{}.png".format(i + 1),
        image.numpy(),
        cmap="gray"
    )
print("16 synthetic images saved successfully.")

# =========================================================
# SAVE GENERATOR MODEL
# =========================================================
generator.save(
    "GAN_MNIST_Generator.keras"
)
print("Generator model saved successfully.")

# =========================================================
# FINAL SYNTHETIC IMAGE
# =========================================================
final_noise = tf.random.normal(
    [1, NOISE_DIM]
)
final_image = generator(
    final_noise,
    training=False
)
plt.figure(figsize=(5, 5))
plt.imshow(
    final_image[0, :, :, 0] * 0.5 + 0.5,
    cmap="gray"
)
plt.axis("off")
plt.title("Final Synthetic Image")
plt.show()
# =========================================================
# FINAL OUTPUT
# =========================================================

print("\nGAN training completed successfully.")
print("Synthetic MNIST images generated successfully.")
print("Student Roll No: 24BAD409")
print("Student Name: Shalini A")
```
